```python
import tensorflow as tf
import numpy as np

# step 1: comvert data to numpy arrays
x = np.array([1, 2, 3, 4], dtype=float)
y = np.array([2, 4, 6, 8], dtype=float)

# step 2: define model using input layer (recommended)
model = tf.keras.Sequential([
    tf.keras.Input(shape=(1,)), #preferred over input_shape in dense
    tf.keras.layers.Dense(1)
])

# step 3: compile the model
model.compile(optimizer='sgd', loss='mean_squared_error')

# step 4: train the model
model.fit(x,y,epochs=10)
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
# convert list to numpy array
x = np.array([1, 2, 3, 4], dtype=float)
y = np.array([2, 4, 6, 8], dtype=float)
```

```python
# Define the model
model = Sequential([
    Dense(1, input_shape=(1,)) # This works, or you can use tf.keras.Input
])
```

```python
#compile the model
model.compile(optimizer='sgd', loss='mean_squared_error')
```

```python
#Train the model
model.fit(x, y, epochs=10)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
```

```python
#define a simple linear model
model = nn.Linear(1, 1)
```

```python
#loss and optimizer
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)
```

```python
#Dummy data
x = torch.tensor([[1,],[2,],[3,],[4.]])
y = torch.tensor([[2,],[4,],[6,],[8.]])
```

```python
# training loop
for epoch in range(10):
  optimizer.zero_grad()
  outputs = model(x)
  loss = criterion(outputs, y)
  loss.backward()
  optimizer.step()
  print(f"Epoch {epoch+1}, Loss: {loss.item()}")
```
